#Creacion de Schemas

```python
%sql
CREATE SCHEMA IF NOT EXISTS Catalogo_Pfinal.bronze;
CREATE SCHEMA IF NOT EXISTS Catalogo_Pfinal.silver;
CREATE SCHEMA IF NOT EXISTS Catalogo_Pfinal.golden;
CREATE SCHEMA IF NOT EXISTS Catalogo_Pfinal.exploratory;
```

#Creacion tablas Bronze

```python
# En Python notebook usando dbutils
dbutils.fs.rm("abfss://bronze@saproyectofinal.dfs.core.windows.net/bronze/CLIENTES", recurse=True)
dbutils.fs.rm("abfss://bronze@saproyectofinal.dfs.core.windows.net/bronze/VENTAS", recurse=True)
dbutils.fs.rm("abfss://bronze@saproyectofinal.dfs.core.windows.net/bronze/PRODUCTOS", recurse=True)
```

```python
%sql
DROP TABLE IF EXISTS  `CLIENTES` ;
CREATE OR REPLACE TABLE Catalogo_Pfinal.bronze.CLIENTES (
  cliente_id STRING,
  nombre STRING,
  edad INT,
  genero STRING,
  ciudad STRING
)
USING DELTA
LOCATION 'abfss://bronze@saproyectofinal.dfs.core.windows.net/bronze/CLIENTES';
DROP TABLE IF EXISTS  `VENTAS` ;
CREATE OR REPLACE TABLE Catalogo_Pfinal.bronze.VENTAS (
  venta_id STRING,
  fecha STRING,
  cliente_id STRING,
  producto_id STRING,
  cantidad INT,
  tienda STRING,
  ciudad STRING
)
USING DELTA
LOCATION 'abfss://bronze@saproyectofinal.dfs.core.windows.net/bronze/VENTAS';
DROP TABLE IF EXISTS  `PRODUCTOS`;
CREATE OR REPLACE TABLE Catalogo_Pfinal.bronze.PRODUCTOS (
  producto_id STRING,
  producto STRING,
  categoria STRING,
  precio DOUBLE
)
USING DELTA
LOCATION 'abfss://bronze@saproyectofinal.dfs.core.windows.net/bronze/PRODUCTOS';
```
